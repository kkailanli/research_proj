```python
from alerce.core import Alerce
import pandas as pd
alerce = Alerce()
```

```python
classifiers = alerce.query_classifiers()
classifiers
```

```python
params = {
   "classifier": "lc_classifier",
   "class_name": "SNII",
   "format": "pandas"
}

page = 1
all_data = []

while True:
   try:
      print(f"Querying page: {page}")
      dataframe = alerce.query_objects(**params,
                                       pagesize = 1000,
                                       page = page)
      if len(dataframe) == 0:
         print("Done")
         break

      all_data.append(dataframe)
      page = page + 1

   except Exception as error:
      print(error)
      break

sn2 = pd.concat(all_data, ignore_index = True)
```
